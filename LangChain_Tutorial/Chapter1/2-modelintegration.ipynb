# Streaming and Batch


### Streaming

Most models can stream thier output content while it is being generated.By displaying output progressively, streaming significantly improves user experience, specially for longer responses.

Calling stream() returns an iterator that yields output chunks as they are produced.
You can use a loop to process each chunk in real-time.

```python
model.invoke("Write me a 200 word paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 200 word paragraph on Artificial Intelligence"):
    print(chunk.text)
```

Using a delimiter to see how each chunk is processed.
Using Flush = True

```python
for chunk in model.stream("Write me a 200 word paragraph on Matrix Movie plot"):
    print(chunk.text, end="|", flush=True)
```

### Batch

Batching a collection of indepenedent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel.

```python
responses = model.batch([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "How is ontology of Knowledge Graph decided?"
])

for response in responses:
    print(response)
```
