# Model Integration with OpenAI, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
```

```python
model = init_chat_model("gpt-4.1")
model
```

```python
## Another way to inoke model

from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-4.1")
model
```

```python
## invoke the model
response = model.invoke("Hello, how are you?")
response
```

# Google Gemini Model Integration

```python
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")


model = init_chat_model("google_genai:gemini-3.8-flash")
```

```python
import requests
import os

api_key = os.environ.get("GOOGLE_API_KEY")
url = f"https://generativelanguage.googleapis.com/v1beta/models?key={api_key}"

response = requests.get(url).json()

# Show only generative text models
for m in response.get('models', []):
    if 'generateContent' in m.get('supportedGenerationMethods', []):
        print(m['name'].replace('models/', ''))
```

```python
response = model.invoke("Why do parrots talk?")
response
```

```python
## Another way to invoke model

import os
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
model

response = model.invoke("Why do humans dream?")
response
```

# GROQ Model Integration

```python
import os

from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-20b")
response = model.invoke("Why do LLM hallucinate?")
response
```

```python
## Another way to invoke a chat model
import os
from langchain_groq import ChatGroq

model = ChatGroq(model="openai/gpt-oss-20b")
response = model.invoke("Why do LLM hallucinate?")
response
```
